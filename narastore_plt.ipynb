```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
dados = {
    'categorias': ['Eletronicos', 'Figures', 'Jogos', 'Mambo'],
    'valores': [4200, 10100, 3600, 280000]
         }
df_dados = pd.DataFrame(dados)
```

```python
plt.figure(figsize=(15, 9)) # Sempre antes de definir o grafico, altera o tamanho da figura
plt.bar(df_dados['categorias'], df_dados['valores']) # qualitativa x quantitativa
plt.title('Total de Vendas por Categoria - NaraStore') 
plt.xticks(rotation=45) # Grau de angulação do eixo X
# plt.tight_layout() serve para impedir a sobreposição de graficos
plt.show()
```

```python
df = pd.read_csv('vendas_narastore.csv')
df
```

```python
df_alimentos_vend = df.groupby('categoria')['total_venda'].sum().reset_index().sort_values(by='total_venda', ascending=False)
df_alimentos_vend
```

```python
plt.barh(df_alimentos_vend['categoria'], df_alimentos_vend['total_venda'])
plt.title('Total de vendas por Categoria - Narastore')
plt.show()
```
