```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dados = {
    'Alunos': ['Ana', 'Bruno', 'Carol', 'Diego', 'Elisa', 'Felipe', 'Gabi', 'Hugo'],
    'Notas': [8.5, 6.0, 9.2, 5.5, 7.8, 4.0, 9.0, 6.5],
    'Horas': [6, 3, 8, 2, 5, 1, 7, 4],
}
df = pd.DataFrame(dados)
df
```

```python
array_hora = np.array(df['Horas'])
array = np.array(df['Notas'])
```

```python
# A unidade da variância sempre sai ao quadrado, nesse caso pontos²
variancia = np.var(array)
print(f"Variância: {variancia:.2f}")
```

```python
desvio = np.std(array)
print(f"Desvio Padrão: {desvio:.2f}")
media = np.mean(array)
print(f"Média: {media:.2f}")
print(f"Faixa Típica: {media-desvio:.2f} até {media+desvio:.2f}")
```

```python
matriz_cov = np.cov(array, array_hora)
print(matriz_cov)
# Coluna e Linha, igual a matriz da escola
cov_valor = matriz_cov[0, 1]
print(f"Covariância: {cov_valor:.2f}")
```

```python
print(df[['Notas', 'Horas']].cov())
```

```python
r = np.corrcoef(array, array_hora)
print(f"Pearson r = {r[0,1]:.4f}")
```

```python
corr_pandas = df[['Notas', 'Horas']].corr()
print(corr_pandas)
```

```python
plt.figure(figsize=(14,6))
plt.scatter(array_hora, array, color='lightgrey', edgecolor='steelblue', s=80)
plt.title('Horas de Estudo vs. Nota')
plt.xlabel('Horas de Estudo')
plt.ylabel('Nota')
plt.show()
```

```python
correlacao = np.corrcoef(array, array_hora)
sns.heatmap(correlacao, 
            annot=True, # Valores dentro das caixas
            cmap='coolwarm', # Cor
            # Valor máximo e mínimo 
            vmin=-1,
            vmax=1,
            # Formatação do valor
            fmt='.2f'
            )
```
