```python
import pandas as pd
import numpy as np

df = pd.read_csv('pedidos_narashop.csv', sep=';')
```

```python
df['valor_total'] = df['quantidade'] * df['valor_unitario']
df
```

```python
# separando o total de cada categoria 
df_categoria_receita = df.groupby('categoria')['valor_total'].sum().reset_index().sort_values(by='valor_total', ascending=False)
df_categoria_receita
```

```python
# Status do pedido
df_entregue_ac500 = df.loc[(df['status_pedido'] == 'Entregue') & (df['valor_total'] > 500)]
df_entregue_ac500
```

```python
df_produtos = df.groupby('produto')['valor_total'].sum().reset_index().sort_values(by='valor_total', ascending=False)
array_produtos_valor_total = np.array(df_produtos['valor_total'])
media_produto = np.mean(array_produtos_valor_total)
mediana_produto = np.median(array_produtos_valor_total)
distancia = round(((media_produto - mediana_produto) /mediana_produto) * 100 ,2)
q1_produtos = np.percentile(array_produtos_valor_total, 25)
q3_produtos = np.percentile(array_produtos_valor_total, 75)
iqr_produtos = q3_produtos - q1_produtos
superior_produtos = q3_produtos + (1.5 * iqr_produtos)
inferior_produtos = q1_produtos - (1.5 * iqr_produtos) # Negativo logo desnecessário 
print(q1_produtos, q3_produtos, superior_produtos)
```

```python
df_menor_q1 = df_produtos.loc[df_produtos['valor_total'] < q1_produtos].sort_values(by='valor_total')
df_menor_q1
```

```python
df_menor_q3 = df_produtos.loc[df_produtos['valor_total'] > q3_produtos]
df_menor_q3
```

```python
outlier_sup_produtos = df_produtos.loc[df_produtos['valor_total'] > superior_produtos]
outlier_sup_produtos
```

```python
df_valor_total_vendedor = df.groupby('vendedor')['valor_total'].sum().reset_index().sort_values(by='valor_total', ascending=False)
array_valor_total_vendedor = np.array(df_valor_total_vendedor['valor_total'])
q1_vendedor = np.percentile(array_valor_total_vendedor, 25)
# q2 = np.percentile(array_valor_total_vendedor, 50)
q3_vendedor = np.percentile(array_valor_total_vendedor, 75)
iqr_vendedor = q3_produtos - q1_produtos
superior_vendedor = q3_vendedor + (1.5 * iqr_vendedor)
inferior_vendedor = q1_vendedor - (1.5 * iqr_vendedor) # Valor negativo logo não conta
print(q1_vendedor, q3_vendedor)
print(superior_vendedor, inferior_produtos)
```

```python
df_vendedor_menor_q1 = df_valor_total_vendedor.loc[df_valor_total_vendedor['valor_total'] < q1_vendedor]
df_vendedor_menor_q1
```

```python
df_vendedor_maior_q3 = df_valor_total_vendedor.loc[df_valor_total_vendedor['valor_total'] > q3_produtos]
df_vendedor_maior_q3
```

```python
outlier_sup_vendedor = df_valor_total_vendedor.loc[df_valor_total_vendedor['valor_total'] > superior_vendedor]
outlier_sup_vendedor
```
