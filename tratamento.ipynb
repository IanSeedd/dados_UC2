```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('dirty/funcionarios_naragroup_sujo.csv', sep=';')
df.info()
```

```python
# count, unique, top e freq 
df.describe(include='object')
```

```python
df.isnull().sum().reset_index()
```

```python
# Percentual de nulos — colunas com >15% exigem atenção especial
(df.isnull().sum() / len(df) * 100).round(1)
# Os valores sairam iguais por coincidencia 
```

```python
# Ver QUAIS linhas têm pelo menos um nulo
df[df.isnull() # Gera True/False para cada dado (True = nulo)
.any(axis=1)] # any(): existe pelo menos um True? axis=1: por linha, axis=0 por coluna
```

```python
# A linha inteira tem que estar duplicada, isso é um problema considerando o ID (não confunda com o index)
df.duplicated().sum()
```

```python
# False marca todas as ocorrências, first só marca a partir da segunda linha e last marca tudo exceto o ultimo
# (No caso do first e false aparece apenas as copias sem repetir o original, o false você veria as copias e o original)
df[df.duplicated(keep=False)]
```

```python
# Duplicata apenas por colunas-chave
df[df.duplicated(subset=['nome','departamento', 'cargo', 'salario', 'data_admissao', 'email', 'ativo'], keep=False)].sort_values(by='nome')
```
