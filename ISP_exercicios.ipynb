```python
import pandas as pd
import numpy as np 
url = 'https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv' 
df = pd.read_csv(url, encoding='latin1', sep=';')
df.columns
```

```python
df_roubo_veic_dp = df.groupby('cisp')['roubo_veiculo'].sum().reset_index()
df_roubo_veic_dp.sort_values(by='roubo_veiculo', ascending= False, inplace=True)

array_roubo_veic_dp = np.array(df_roubo_veic_dp['roubo_veiculo'])
media = np.mean(array_roubo_veic_dp)
mediana = np.median(array_roubo_veic_dp)
q1 = np.percentile(array_roubo_veic_dp, 25)
q2 = np.percentile(array_roubo_veic_dp, 50)
q3 = np.percentile(array_roubo_veic_dp, 75)

df_roubo_veic_dp_acima_q3 = df_roubo_veic_dp.loc[df_roubo_veic_dp['roubo_veiculo'] > q3]
#df_roubo_veic_dp_acima_q3
distancia = (media - mediana) / mediana # Número percentual por isso sai um decimal, depois é só formatar para %
# A distancia é puxada pelos valores extremos que afetam a media para cima ou para baixo, só lembrar das relações matemáticas. Literalmente pura matemática.
print(f'Média: {media}\nMediana: {mediana} \nDistânca: {distancia:.2%}') # Seja rigoroso com a distancia mas em projetos de portfolio use o letramento tecnico padrão
```

```python
df_baixada = df.loc[df['regiao'] == 'Baixada Fluminense'] # Baixa Fluminense != Baixada
df_roubo_veic_dp_baixada = df_baixada.groupby('cisp')['roubo_veiculo'].sum().reset_index()
df_roubo_veic_dp_baixada.sort_values(by='roubo_veiculo', ascending= False, inplace=True)

array_roubo_veic_dp_baixada = np.array(df_roubo_veic_dp_baixada)
q1 = np.percentile(array_roubo_veic_dp_baixada, 25)
q2 = np.percentile(array_roubo_veic_dp_baixada, 50)
q3 = np.percentile(array_roubo_veic_dp_baixada, 75)

df_roubo_veic_dp_baixada = df_roubo_veic_dp_baixada.loc[df_roubo_veic_dp_baixada['roubo_veiculo'] > q3]
df_roubo_veic_dp_baixada
```

```python
df_ap_menores_dp = df.groupby('cisp')['aaapai'].sum().reset_index()
df_ap_menores_dp.sort_values(by='aaapai', ascending= False, inplace=True) 

array_ap_menores_dp = np.array(df_ap_menores_dp)
media = np.mean(array_ap_menores_dp)
mediana = np.median(array_ap_menores_dp)
distancia = (media - mediana) / mediana

print(f'Média: {media} \nMediana: {mediana} \nDistânca: {distancia:.2%}')
```
